```python
import numpy as np
import matplotlib.pyplot as plt

from lichess_client import fetch_games, build_rating_series
from analytics import fit_line, predict
```

```python
username = "Kittens_the_first"
game_type = "rapid"
games_ahead = 10

raw_games = fetch_games(username, game_type)
ratings = build_rating_series(raw_games, username, speed=game_type)
model = fit_line(ratings)
print(len(ratings), model["slope"], model["r_squared"], predict(model, games_ahead))
```

```python
x = np.arange(len(ratings))
x_future = np.arange(len(ratings) - 1, len(ratings) + games_ahead)

trend = model['slope'] * x + model["intercept"]
forecast = model["slope"] * x_future + model["intercept"]

plt.figure(figsize=(10, 5))
plt.scatter(x, ratings, s=18, label="games")
plt.plot(x, trend, color="tab:red", label="trend")
plt.plot(x_future, forecast, color="tab:red", linestyle="--",
         label=f"forecast (+{games_ahead} games)")
plt.title(f"{username}: {game_type} rating trend")
plt.xlabel("game number")
plt.ylabel("rating")
plt.grid(alpha=0.3)
plt.xticks(np.arange(0, len(ratings) + games_ahead + 10, 10))
plt.legend()
plt.show()
```
